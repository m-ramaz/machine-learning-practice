```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
array1=np.linspace(0,5,11)   #ye function evenly spaced numbers ki list return krta hai.
```

```python
array1
```

```python
array2=np.linspace(0,5,11,endpoint=False)    # np.linspace(star,end,numbers jitny chahye,agr endpoint chahye =true wrna fales)
```

```python
array2
```

```python
array_for_graph=array1**2
```

```python
array_for_graph
```

```python
plt.plot(array1,array_for_graph)   #ye function sirf line graph bnata hai simple default setting ke sth
```

```python
plt.plot()   #ye function simple canvas bnata hai
```

# **Sub-PLots**

```python
plt.subplot(2,2,1)
plt.plot(array1,array_for_graph,'r')
plt.title('Plot No 1')
plt.subplot(2,2,2)
plt.plot(array1,array_for_graph,'b')
plt.title('Plot No 2')
plt.subplot(2,2,3)
plt.plot(array1,array_for_graph,'g')
plt.title('Plot No 3')
plt.subplot(2,2,4)
plt.plot(array1,array_for_graph,'y')
plt.title('Plot No 4')
```

```python

plt.subplot(3,3,1)
plt.plot(array1,array_for_graph,'r')
plt.title('Plot No 1')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.subplot(3,3,2)
plt.plot(array1,array_for_graph,'b')
plt.title('Plot No 2')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.tight_layout()  # ye graphs ko sahi se alag alag kr ke bnata hai.
```

```python
plt.subplot(2,2,1)
plt.subplot(2,2,2)
plt.subplot(2,2,3)
```

```python
fig=plt.figure()   # aik window object bnay ga
axes1=fig.add_axes([0.2,0.8,0.4,0.4])   # is se hum aik axes yani graph bnaye gay

axes2=fig.add_axes([0.2,0.4,0.4,0.4])  #is se dosra graph bny ga

axes3=fig.add_axes([0.4,0.4,0.2,0.2])                  #fig.add_axes()  iske indr list pass kro pehle 2 arguments left se aur nechy se khiskana
```

```python
fig=plt.figure(figsize=(8,8),dpi=100)   #dpi pixel ko dekhati hai   figsize -> width aur height ki value di jati window ke lye.
axes1=fig.add_axes([0.1,0.1,0.8,0.8])
plt.plot(array1,array_for_graph,'r')
plt.title('Plot No 1')
```

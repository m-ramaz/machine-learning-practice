```python
import pandas as pd
import numpy as np
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.read_csv("/content/covid_toy.csv")
```

```python
df.drop(columns=['has_covid'])
```

```python
x_train,x_test,y_train,y_test=train_test=train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2,random_state=0)
```

# **one hot encoding**

```python
# to apply one hot encoding we will use class of OneHotEncoder
ohe=OneHotEncoder(drop='first',sparse_output=False)
x_train_gender_city=ohe.fit_transform(x_train[['gender','city']])
# testing data pr bhi lagay ga
x_test_gender_city=ohe.transform(x_test[['gender','city']])
```

# **OrdinalEncoding**

```python
# apply ordinal encoding on ordinal categorical data
oe=OrdinalEncoder()
x_train_cough=oe.fit_transform(x_train[['cough']]) # ye transformation array return krti hai.
x_test_cough=oe.transform(x_test[['cough']])
```

```python
x_train_cough.shape
# x_train  ye aik array return krta hai.
```

# **SimpleImputer**

```python
si=SimpleImputer()
# fever ki missing values hai isle wahan ye apply hu ga.
x_train_fever=si.fit_transform(x_train[['fever']])
x_test_fever=si.transform(x_test[['fever']])
```

```python
x_train_fever.shape  # ab dekho yahan bhi values 80 rows show kr rha.
```

```python
# ab ye tamam transformation mene alag alag apply ki hai lekin ab mujhy in sab ko ikhata krna pry ga takeh wo
# x_train aur x_test ki tarha pora ka pora aik sth dataframe nzr aye.
```

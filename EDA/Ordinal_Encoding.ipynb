```python
# ordinal encoding unn columns ki hoti hai jo categorical hu aur un me kuch order hu
```

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("/content/customer.csv")
df.head()
```

```python
df_new=df.iloc[:,2:4]
df_new.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(df_new,df['purchased'],test_size=0.3,random_state=0)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
df.shape
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe=OrdinalEncoder(categories=[['Good','Average','Poor'],['School','UG','PG']])
x_train_scaled=oe.fit_transform(x_train)
x_test_scaled=oe.transform(x_test)
```

```python
x_train_scaled_df=pd.DataFrame(x_train_scaled,columns=x_train.columns)
x_test_scaled_df=pd.DataFrame(x_test_scaled,columns=x_test.columns)
```

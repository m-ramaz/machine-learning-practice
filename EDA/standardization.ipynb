# splitting Data

```python
X=df.drop("Purchased",axis=1)
X
Y=df["Purchased"]
Y
```

**Train test split**

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.3,random_state=0)
```

```python
x_test
```

```python
y_train
y_test
```

```python
x_train.shape
x_test.shape
```

# **Scaling training data**

```python
from sklearn.preprocessing import StandardScaler
# sklearn library se aik module hai preprocessing usme me se StandardScaler aik
# class hai wo import ki gai hai
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
x_train_scaled   # ye aik array return krti hai but we want dataframe
```

```python
x_train_scaled=pd.DataFrame(x_train_scaled,columns=x_train.columns)
x_train_scaled
# jo numpy array return hoi hai usko dubara dataframe me convert kia hai aur
# jo parameter columns wala hai wo columns ke naam le rha hai pehle waly
# unscale dataframe se
```

```python
x_test_scaled=pd.DataFrame(x_test_scaled,columns=x_test.columns)
```

```python
x_test_scaled
```

# **why was scaling importent let see the difference**

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr_scaled=LogisticRegression()
```

```python
lr.fit(x_train,y_train)
lr_scaled.fit(x_train_scaled,y_train)
```

**now we do prediction on scaled and without scaled values**

```python
y_pred=lr.predict(x_test)
y_pred_scaled=lr_scaled.predict(x_test_scaled)
```

**Accuracy check**

```python
from sklearn.metrics import accuracy_score
print("Actual",accuracy_score(y_test,y_pred))
print("Scaled",accuracy_score(y_test,y_pred_scaled))
```

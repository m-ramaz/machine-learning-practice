```python
import pandas as pd
import scipy.stats as stats

# 1. Target aur Categorical columns define karein
target = 'charges'
cat_cols = ['smoker', 'sex_encoded', 'region_northwest', 'region_southeast', 'region_southwest', 'bmi_category']

target_results = []
alpha = 0.05  # Significance level

# 2. Har categorical column ka charges ke sath test karein
for col in cat_cols:
    # Har category ke mutabiq charges ka group banayein (e.g., smoker=0 ke charges vs smoker=1 ke charges)
    groups = [df1[df1[col] == val][target] for val in df1[col].unique()]

    # Agar column me sirf 2 groups hain (e.g., 0 and 1) -> T-Test lagayein
    if len(groups) == 2:
        test_name = 'T-Test'
        stat, p_value = stats.ttest_ind(groups[0], groups[1], equal_var=False)
    # Agar 2 se zyada groups hain (e.g., bmi_category 1, 2, 3) -> ANOVA Test lagayein
    elif len(groups) > 2:
        test_name = 'ANOVA'
        stat, p_value = stats.f_oneway(*groups)
    else:
        continue  # Agar column me data hi na ho

    # Decision logical condition
    if p_value < alpha:
        decision = "Reject H0 (Highly Important for Prediction)"
    else:
        decision = "Accept H0 (Not Important / Insignificant)"

    # Results ko list me save karna
    target_results.append({
        'Feature Column': col,
        'Test Used': test_name,
        'P-Value': round(p_value, 6),
        'Decision': decision
    })

# 3. Final Result ko proper DataFrame Table format me convert karein
target_analysis_df = pd.DataFrame(target_results)

# Table ko descending order me sort karein (sab se important columns upar aayenge)
target_analysis_df = target_analysis_df.sort_values(by='P-Value')

print(target_analysis_df)
```

# **Model Training and Testing**

```python
#importing
from sklearn.model_selection import train_test_split
```

```python
x=final_df.drop('charges',axis=1)
y=final_df['charges']
```

```python
# Split into 80% training and 20% testing
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
print(f"Training features shape: {x_train.shape}")
print(f"Testing features shape: {x_test.shape}")
print(f"Training features shape: {y_train.shape}")
print(f"Testing features shape: {y_test.shape}")
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model_lr=LinearRegression()
```

```python
model_lr.fit(x_train,y_train)
```

```python
y_pred=model_lr.predict(x_test)  # my model has predicted values against the testing data
```

```python
#to check how much model has learned or is strong or trained good or bad.
from sklearn.metrics import r2_score
```

```python
r2=r2_score(y_test,y_pred)   #80% model is predicting right.
r2
```

```python
#adjusted r2 score is used to check that is our features are good enough for prediction.
n=x_test.shape[0]   #this gives rows of x_test dataframe
p=x_test.shape[1]   #this gives the columns count
adjusted_r2=1-((1-r2)*(n-1)/(n-p-1))
adjusted_r2                            #the score is almost same so the model is in good situation.
```
